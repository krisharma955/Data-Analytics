## Advance Operations on Data

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("==== Zomato Sales Analysis ====")
print("\n Sales Data Shape: ", sales_data.shape)
print("\n Sample Data for 1st Three Restaurants: \n", sales_data[:3])
print("\n Sample Data without IDs: \n", sales_data[:, 1:])
```

```python
#total sales per year
yearly_sales = np.sum(sales_data[:, 1:], axis=0)
print(yearly_sales)
```

```python
#Minimum Sales per Restaurants
min_sales = np.min(sales_data[:, 1:], axis=1)
print("Minimum Sales per Restaurants: ",min_sales)

#Minimum Sales Per Year
min_sales_year = np.min(sales_data[:, 1:], axis=0)
print("\n Minimum Sales Per Year: ", min_sales_year)

#Maximum Sales Per Year
max_sales_year = np.max(sales_data[:, 1:], axis=0)
print("\n Maximum Sales Per Year: ", max_sales_year)

#Average Sales Per Restaurants
avg_sales_res = np.mean(sales_data[:, 1:], axis=1)
print("\n Average Sales Per Restaurants: ", avg_sales_res)
```

```python
#Cumulative Sum
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(cumsum)

plt.figure(figsize = (8, 6)) 
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average Cumulative Per Year Across all Restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

### Vector Operations

```python
v1 = np.array([1, 2, 3, 4, 5])
v2 = np.array([6, 7, 8, 9, 10,])

print("Vector Addition: ", v1 + v2)
print("\n Vector Multiplication: ", v1 * v2)
print("\n Vector Dot Product: ", np.dot(v1, v2))

angle = np.arccos(np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2)))
print("Angle: ", angle)
```

```python
#VECTORIZED OPERATIONS

restaurant_types = np.array(["biryani", "chinese", "indian", "thai"])
vectorized_upper = np.vectorize(str.upper) #regular py method converted into vectorized operation
print("Vectorized Upper: ", vectorized_upper(restaurant_types))
```
