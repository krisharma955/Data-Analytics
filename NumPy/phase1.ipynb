## NumPy Arrays & Basics

```python
import numpy as np
```

### Creating Array from List

```python
arr_1d = np.array([1, 2, 3])
print("1D Array: ", arr_1d)

arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("2D Array: ", arr_2d)
```

### List vs NumPy Array

```python
py_list = [1, 2, 3]
print("Py List Multi: ", py_list * 2)

np_array = np.array([1, 2, 3]) #element wise multiplication
print("Py Array Multi: ", np_array * 2)
```

```python
import time 

start = time.time()
py_list = [i*2 for i in range(100000)]
print("py list operation time: ", time.time() - start)

start = time.time()
np_array = np.arange(100000) * 2 #numpy computation time is much faster 
print("np array operation time: ", time.time() - start)
```

### Creating Array from Scratch

```python
zeros = np.zeros((3, 4)) #to create a zeros array
print("zeros array: \n", zeros)

ones = np.ones((4, 5)) #to create a ones array
print("ones array: \n", ones)

full = np.full((2, 2), 8) #to create a constants array
print("full array: \n", full)

random = np.random.random((2, 5)) #random class -> random() -> to create a random array - values: [0, 1]
print("random array: \n", random)

sequence = np.arange(0, 11, 2) #(start, stop, step) - stop (not included)
print("sequence array: \n", sequence)
```

### Vector, Matrix & Tensor

```python
vector = np.array([1, 2, 3])
print("Vector: ", vector)

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
print("Matrix: ", matrix)

tensor = np.array([[[1, 2], [3, 4]], #multi-dimensional arrays
                   [[5, 6], [7, 8]]])
print("Tensor: ", tensor)
```

### Array Properties

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Shape: ", arr.shape) #rows & cols
print("Dimensions: ", arr.ndim) #dimensions
print("Size: ", arr.size) #no of elements
print("Data type: ", arr.dtype) #data type of elements 
```

### Array Reshaping

```python
arr = np.arange(12) #non-exclusive range
print("org arr: ", arr)

reshaped = arr.reshape((3, 4)) #reshaping into rows & cols
print("reshaped arr: \n", reshaped)

flatten = reshaped.flatten() #flatting - single row
print("flatten arr: ", flatten)

#ravel -> (returns view, instead of copy)
raveled = reshaped.ravel()
print("raveld arr: ", raveled)

#transpose - rows -> cols | cols -> rows
transpose = reshaped.T
print("transposed arr: \n", transpose)
```
