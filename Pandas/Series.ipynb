## Series

Panda Series -> 1D labeled array capable of holding any data type. -> Vertical

### Creating Series

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a', 'b', 'c']
my_list = [10, 20, 30]
arr = np.array([10, 20 ,30])
d = {1: 10, 2: 20, 3: 30}
```

```python
#List
pd.Series(my_list)
```

```python
#List (w custom idxs)
pd.Series(my_list, index = labels)
```

```python
#NumPy Array
pd.Series(arr) #index can also be used here
```

```python
#Dictionary
pd.Series(d)
```
