## DataFrames

When you combine multiple Series together, it forms a Dataframe.

```python
import numpy as np
import pandas as pd
```

### Creating a DataFrame

```python
#Using Dictionary

data = {
    'Name': ['John', 'Peter', 'Mark'],
    'Age': [28, 42, 32],
    'City': ['NYC', 'LA', 'CAL'],
    'Salary': [45000, 55000, 85000]
}

df_dict = pd.DataFrame(data)
```

```python
df_dict
```

```python
#Using List

data_list = [
    ['John', 28, 'NYC', 65000],
    ['Peter', 26, 'LA', 75000],
    ['Mark', 45, 'CAL', 85000],
]

df_list = pd.DataFrame(data_list)

#Colums (headers)
cols = ['Name', 'Age', 'City', 'Salary']
df_list_col = pd.DataFrame(data_list, columns = cols)
```

```python
df_list
```

```python
df_list_col
```

### Selection & Indexing of Columns

```python
df_list_col
```

```python
df_list_col['Name']
```

```python
df_list_col['City']
```

```python
#Multiple Cols at Once

df_list_col[['Name', 'Age']]
```

### Creating a new column

```python
df_list_col
```

```python
df_list_col["Designation"] = ["Doctor", "Engg", "CA"]
df_list_col
```

### Removing Columns

```python
# doesn't permanantly removes cols / rows gives a copy of if deleted
df_list_col.drop("Designation", axis = 1)
```

```python
df_list_col #Designation not dropped
```

```python
df_list_col.drop("Designation", axis = 1, inplace = True) #Permanantly Delete- (inplace = True)
```

```python
df_list_col
```

```python
#Removing Multiple Colums

df_list_col.drop(["City", "Salary"], axis = 1, inplace = True)
```

```python
df_list_col
```

```python
# Removing Rows -> (axis = 0)

df_list_col.drop(0, axis = 0)
```

### Selecting Row

```python
# loc (location) - for rows

df_list_col.loc[0]
```

```python
df_list_col.loc[[0, 1]]
```

```python
# iloc (index location)

df_list_col.iloc[2]
```

### Selecting Subsets of Rows & Cols

```python
df_list_col.loc[[0, 1]][["Age", "Designation"]]
```

```python
df_list_col.loc[[1, 2]][["Name", "Age"]]
```

### Conditional Selection

```python
df_list_col
```

```python
# People with Age > 26
df_list_col[df_list_col["Age"] > 26]
```

```python
#People with Age > 26 & Designation CA
df_list_col[(df_list_col["Age"] > 26) & (df_list_col["Designation"] == "CA")]

#for OR -> | 
```
