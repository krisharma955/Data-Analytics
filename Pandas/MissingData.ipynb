## Missing Data

##### Empty Values in DataSet -> either fill them or remove them

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

df = pd.DataFrame(data)
```

```python
df
```

### Finding Missing/NULL Data

```python
df.isna()
```

```python
df.isna().sum() #for count of null values - col wise
```

```python
df.isna().any() #for checking if each col has null value or not
```

### Removing Missing/NULL Data

```python
df
```

```python
# Removing rows containing NULL Value

df.dropna()
```

```python
# thresh - threshold for non-null values - in rows

df.dropna(thresh = 3) #atleast 3 non-null values 
```

### Filling the Missing Data

```python
#Filling All Null Values with 0

df.fillna(0)
```

```python
#Filling each col with different values - col wise

vals = {
    'A': 0,
    'B': 100,
    'C': 300,
    'D': 500
}

df.fillna(value = vals)
```

```python
#Filling with mean

df.fillna(df.mean())
```
